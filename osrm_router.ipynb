```python
import requests
```

```python
class OSRMRouter:

    def __init__(self):
        self.base_url = "https://router.project-osrm.org"

    def get_route(self, start, end):

        url = (
            f"{self.base_url}/route/v1/driving/"
            f"{start[0]},{start[1]};{end[0]},{end[1]}"
        )

        params = {
            "overview": "full",
            "geometries": "geojson"
        }

        response = requests.get(url, params=params)
        data = response.json()

        if data["code"] != "Ok":
            return None

        route = data["routes"][0]

        return {
            "distance": route["distance"] / 1000,
            "duration": route["duration"] / 60,
            "coordinates": route["geometry"]["coordinates"]
        }
```

```python
if __name__ == "__main__":

    router = OSRMRouter()

    start = (75.8513, 26.9855)      # Amber Fort
    end = (75.8267, 26.9239)        # Hawa Mahal

    result = router.get_route(start, end)

    print(result)
```
